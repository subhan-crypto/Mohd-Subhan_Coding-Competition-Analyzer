## 3. Explore the data

```python
# TODO: Display the shape and column names
print("shape(rows,colums):", df.shape)
print("Column names:",list(df.columns))

print("Data information:", df.info())

# TODO: Inspect data types and missing values
print("Missing values:", df.isnull().sum())

# TODO: Count duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project

# TODO: Convert them to a NumPy array using .to_numpy()
score_array = df["Score"].to_numpy()
solved_array = df["Problems_Solved"].to_numpy()
# TODO: Calculate average, maximum and minimum
print("Average score:", np.mean(score_array))
print("Highest score:", np.max(score_array))
print("Lowest score:", np.min(score_array))
print("Average problems solved:", np.mean(solved_array))

```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
df["Unsolved_Problems"] = df["Problems_Attempted"] - df["Problems_Solved"]
df["Solve_Rate"] = (df["Problems_Solved"] / df["Problems_Attempted"].replace(0, np.nan) * 100).fillna(0).round(2)
print("Students with a score of 30 or more:")
display(df[df["Score"] >= 30])

sorted_students = df.sort_values(by="Score", ascending=False)
print("Students sorted by score:")
display(sorted_students)
# TODO: Create calculated column(s)

# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
```

## 6. Your individual project challenge
**Challenge:** Calculate solve rate; create leaderboard; identify top five scorers.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# highest score gets rank 1.
leaderboard = df.sort_values(by=["Score", "Problems_Solved"],ascending=[False, False]).copy()
leaderboard["Leaderboard_Rank"] = range(1, len(leaderboard) + 1)
# complete leaderboard.
print("Complete leaderboard:")
display(leaderboard[["Leaderboard_Rank", "Student_ID", "Name", "Score","Problems_Attempted", "Problems_Solved", "Solve_Rate"]])
# the top five scorers.
top_five = leaderboard.head(5)
print("Top five scorers:")
display(top_five[["Leaderboard_Rank", "Student_ID", "Name", "Score","Problems_Attempted", "Problems_Solved", "Solve_Rate"]])
```
